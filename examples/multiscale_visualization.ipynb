# Multiscale Elastic Time Warping: one recorded run

This notebook visualizes the actual coarse-to-fine implementation. Blue cells are evaluated by the dynamic program; the orange line is the selected matching. The example uses a fixed radius so that the geometry is easy to see.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if ROOT.name == 'examples':
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from examples.multiscale_visualization import record_fixed_radius_run, plot_trace
```

## Run used for the presentation illustration

The hierarchy has sizes $32,64,128,256$.  The coarsest grid is solved in full. At each finer level, the preceding matching is lifted and dilated by four indices.

```python
trace = record_fixed_radius_run(
    n=256,
    seed=356,
    noise=0.02,
    warp_strength=0.5,
    sigma=0.25,
    coarsest_size=32,
    radius=4,
)
trace.summary_rows()
```

```python
fig = plot_trace(trace)
fig
```

```python
print(f'Total evaluated cells across all levels: {trace.total_allowed_cells:,}')
print(f'Cells in the finest full grid:          {trace.finest_full_cells:,}')
print(f'Reduction factor:                       {trace.finest_full_cells / trace.total_allowed_cells:.2f}x')
print(f'Fixed-radius final score:                {trace.final_score:.15f}')
print(f'Unrestricted exact score:                {trace.exact_score:.15f}')
print(f'Score gap:                               {trace.exact_score - trace.final_score:.3g}')
```

## Interpretation

Each restricted solve is exact **inside its shaded corridor**.  In this recorded run the final score equals the unrestricted exact score, but fixed-radius multiscale alignment is not certified exact in general. The adaptive implementation widens the radius when consecutive scores are unstable or the selected path touches an artificial band boundary.
